```python
 %%writefile persegipanjang.cpp
#include <iostream>
using namespace std;

int main() {
    int P, L, Lk;

    cout << "Masukkan panjang (P): ";
    cin >> P;
    cout << "Masukkan lebar (L): ";
    cin >> L;

    Lk = 2 * (P + L);

    cout << "Keliling persegi panjang: " << Lk << endl;
    return 0;
}
```
